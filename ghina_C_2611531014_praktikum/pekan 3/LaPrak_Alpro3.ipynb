## 2.1 Dasar Teori
1. Operator Aritmatika
Operator aritmatika merupakan operator yang digunakan untuk melakukan perhitungan matematika dalam Python. Operator yang digunakan antara lain + untuk penjumlahan, - untuk pengurangan, * untuk perkalian, / untuk pembagian, // untuk pembagian bilangan bulat, % untuk mencari sisa pembagian, dan ** untuk perpangkatan. Dalam program transaksi, operator aritmatika dapat digunakan untuk menghitung diskon, total pembayaran, dan harga rata-rata barang.

2. Operator Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari perbandingan berupa True atau False. Beberapa operator perbandingan yaitu ==, !=, >, <, >=, dan <=. Dalam program transaksi, operator ini dapat digunakan untuk memeriksa apakah total belanja sudah mencapai Rp200.000 atau apakah jumlah barang sudah memenuhi minimal 3 barang.

3. Operator Logika
Operator logika digunakan untuk menggabungkan atau membalik beberapa kondisi. Operator and menghasilkan True jika semua kondisi terpenuhi, or menghasilkan True jika salah satu kondisi terpenuhi, sedangkan not digunakan untuk membalik hasil suatu kondisi. Dalam program transaksi, operator logika dapat digunakan untuk menentukan apakah pelanggan mendapatkan diskon berdasarkan status member, jumlah belanja, atau ketersediaan promo.

4. Operator Bitwise
Operator bitwise merupakan operator yang digunakan untuk melakukan operasi terhadap bilangan berdasarkan bentuk biner atau bit. Beberapa operator bitwise yang digunakan adalah & untuk AND, | untuk OR, ^ untuk XOR, << untuk menggeser bit ke kiri, dan >> untuk menggeser bit ke kanan. Dalam program transaksi, operator bitwise dapat digunakan untuk membuat kode status pelanggan berdasarkan beberapa kondisi, seperti status member, jumlah belanja, jumlah barang, dan ketersediaan kode promo.

```python
# Buat file dengan nama aritmatika_2611531014.py
# Buat program untuk operator aritmatika dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1014
# program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_1014 = int(input("input angka-1: "))
angka2_1014 = int(input("input angka-2: "))

# Penjumlahan
hasil = angka1_1014 + angka2_1014
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

# Pengurangan 
hasil = angka1_1014 - angka2_1014
print("\nOperator Pengurangan")
print("Hasil =", hasil)

# Perkalian 
hasil = angka1_1014 * angka2_1014
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_1014 != 0:
    hasil = angka1_1014 / angka2_1014
    print("\nOperator Pembagian")
    print("Hasil=", hasil)

    hasil = angka1_1014 // angka2_1014
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_1014 % angka2_1014
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil = angka1_1014 ** angka2_1014
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

Jika terpenuhi, program menghitung pembagian desimal (/), pembagian bulat (//), dan sisa bagi/modulo (%). Jika bernilai nol, blok else akan menampilkan pesan peringatan.

### 2.2.2 Program "C_2611531014"

```python
# Buat file dengan nama Assingment_2611531014.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1014
# program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam python

angka1_1014 = int(input("Input angka-1: "))
angka2_1014 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_1014)
print("Nilai angka2 =", angka2_1014)

# Assingment biasa 
hasil = angka1_1014
print("\nAssignment baisa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_1014
hasil += angka2_1014
print("nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assingment pengurangan
hasil = angka1_1014
hasil -= angka2_1014
print("nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assingment perkalian
hasil = angka1_1014
hasil *= angka2_1014
print("nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assingment pembagian, pembagian bulat, dan sisa bagi
if angka2_1014 != 0:
    hasil = angka1_1014
    hasil /= angka2_1014
    print("nAssignment pembagian (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1_1014
    hasil //= angka2_1014
    print("nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_1014
    hasil %= angka2_1014
    print("nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)     
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1_1014
hasil **= angka2_1014
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil)
```

**Penjelasan:**
1. Komentar & Petunjuk: Baris # berisi dokumentasi program, nama file Assingment_2611531014.py, dan aturan variabel bertag NIM 1014.

2. Input & Konversi Data: Perintah int(input(...)) mengambil input pengguna lalu mengubah tipe data string menjadi integer untuk disimpan di angka1_1014 dan angka2_1014.

3. Menampilkan Nilai Awal: Perintah print() menampilkan nilai dasar kedua variabel sebelum diproses lebih lanjut.

4. Assignment Dasar & Operasi Aritmatika (=, +=, -=, *=): Variabel hasil diisi awal dengan = lalu dihitung secara ringkas menggunakan penugasan penjumlahan (+=), pengurangan (-=), dan perkalian (*=).

5. Percabangan & Assignment Pembagian (/=, //=, %=): Kondisi if angka2_1014 != 0 mengecek agar pembagi bukan nol, lalu menjalankan assignment pembagian biasa, pembagian bulat, dan modulo. Jika nol, pesan peringatan else akan muncul.

6. Assignment Perpangkatan (**=): Menghitung perpangkatan variabel hasil dengan eksponen angka2_1014 menggunakan operator **= lalu mencetak hasilnya.

### 2.2.3 Program "C_2611531014"

```python
# Buat file dengan nama bitwise_2611531014.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1014
# program ini menggunakan fungsi input()

print("\n============================")
print("3. OPERATOR BITWISE")
print("==============================")

angka1_1014 = int(input("Masukkan angka bitwise-1: "))
angka2_1014 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_1014, "| biner =", bin(angka1_1014))
print("angka2 =", angka2_1014, "| biner =", bin(angka2_1014))

# Bitwise AND
hasil = angka1_1014 & angka2_1014
print("\nBitwise AND (&)")
print(angka1_1014, "&", angka2_1014, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit)=", format(hasil, "08b"))

# Bitwise OR
hasil = angka1_1014 | angka2_1014
print("\nBitwise OR (|)")
print(angka1_1014, "|", angka2_1014, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit)=", format(hasil, "08b"))

# Bitwise XOR
hasil = angka1_1014 ^ angka2_1014
print("\nBitwise XOR (^)")
print(angka1_1014, "^", angka2_1014, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit)=", format(hasil, "08b"))

# Bitwise NOT
hasil = ~angka1_1014
print("\nBitwise NOT (~)")
print("~", angka1_1014, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit)=", format(hasil, "08b"))

# Bitwise geser kiri
jumlah_geser_1014 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil = angka1_1014 << jumlah_geser_1014
print("\nBitwise geser kiri (<<)")
print(angka1_1014, "<<", jumlah_geser_1014, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit)=", format(hasil, "08b"))

# Bitwise geser kanan
hasil = angka1_1014 >> jumlah_geser_1014
print("\nBitwise geser kanan (>>)")
print(angka1_1014, ">>", jumlah_geser_1014, "=", hasil)
print("Biner hasil =", bin(hasil))
print("Biner hasil (8 bit)=", format(hasil, "08b"))
```

**Penjelasan:**
1. Komentar & Input Data: Baris awal berfungsi menerima input dua bilangan desimal dari pengguna via int(input()) yang disimpan pada variabel angka1_1014 dan angka2_1014 berakhiran NIM.

2. Konversi & Tampilan Biner Awal: Menggunakan fungsi bin() untuk mengonversi nilai desimal variabel ke format biner lalu mencetaknya ke layar sebagai acuan perbandingan.

3. Operasi Bitwise Logika (&, |, ^, ~): Masing-masing bagian ini melakukan operasi logika bit-demi-bit (AND, OR, XOR, dan NOT/Negasi) pada dua nilai. Hasil desimal dan bentuk biner 8-bit (format(hasil, "08b")) langsung dicetak.

4. Input Jumlah Pergeseran: Perintah input() kedua meminta pengguna memasukkan jumlah nilai pergeseran posisi bit yang disimpan dalam jumlah_geser_1014.

5. Operasi Bitwise Shift (<<, >>): Melakukan pergeseran bit ke kiri (<<) untuk mengalikan nilai dan pergeseran bit ke kanan (>>) untuk membagi nilai berbasis 2, kemudian menampilkan nilai desimal serta format biner 8-bit hasilnya.

### 2.2.4 Program "C_2611531014"

```python
# Buat file dengan nama lainnya_2611531014.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1014
# program ini menggunakan fungsi input()
# Program operator keanggotaan dan indentitas

print("=================================")
print("1. OPERATOR KEANGGOTAAN")
print("=================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_1014 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_1014= [int(angka_1014.strip()) for angka_1014 in input_data_1014.split(",")]

nilai_dicari_1014 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_1014= nilai_dicari_1014 in data_1014
print("\nOperator keanggotaan IN")
print(nilai_dicari_1014, "in", data_1014, "=", hasil_1014)

# Operator not in
hasil_1014 = nilai_dicari_1014 not in data_1014
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_1014, "in", data_1014, "=", hasil_1014)


print("=================================")
print("2. OPERATOR IDENTITAS")
print("=================================")

# objek1 menggunakan list dari input pengguna
objek1_1014 = data_1014

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1014 = objek1_1014

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1014 = data_1014.copy()

print("objek1 =", objek1_1014)
print("objek2 =", objek2_1014)
print("objek3 =", objek3_1014)

# Operator is 
hasil = objek1_1014 is objek2_1014
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil)

# Operator is not
hasil = objek1_1014 is not objek3_1014
print("\nOperator identitas IS NOT")    
print("objek1 is not objek3 =", hasil)

# Membandingkan identitas dan nilai
print("\nMembandingkan identitas dan nilai")
print("objek1 is objek3 =", objek1_1014 is objek3_1014)
print("objek1 == objek3 =", objek1_1014 == objek3_1014)
```

**Penjelasan:**
1. Pemrosesan Input List, meminta deretan angka berpisah koma, mengonversinya menjadi list integer (data_1014), lalu mengambil angka acuan (nilai_dicari_1014).

2. Operator Keanggotaan (in & not in), memeriksa ada atau tidaknya angka di dalam list; in menghasilkan True jika ada, sedangkan not in menghasilkan True jika tidak ada.

3. Inisialisasi Memori Objek, objek2_1014 dibuat merujuk lokasi memori yang sama dengan objek1_1014, sedangkan objek3_1014 disalin via .copy() sehingga menghasilkan objek baru dengan isi yang sama.

4. Operator Identitas (is & is not), membandingkan apakah dua variabel menunjuk ke alamat memori yang sama (is) atau beda (is not).

5. Perbandingan Identitas (is) vs Nilai (==), membuktikan bahwa is mengecek kesamaan memori (False untuk objek1 & objek3), sedangkan == mengecek kesamaan isi data (True).

### 2.2.5 Program "C_2611531014"

```python
# Buat file dengan nama logika_2611531014.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1014
# program ini menggunakan fungsi input()
# Program operator logika dalam python

# Memasukkan nilai boolean 
# Input tidak peka terhadap huruf besar dan kecil
a1_1014 = input("Input nilai booelan-1 (true/false): ").strip().lower() == "true"
a2_1014 = input("Input nilai booelan-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_1014)
print("A2 =", a2_1014)

# Konjungsi: bernilai True jika keduanya True
hasil = a1_1014 and a2_1014
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_1014 or a2_1014
print("\nDijungsi (OR)")
print("A1 or A2 =",hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_1014
print("\nNegasi A2 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_1014
print("\nNegasi A2 (NOT)")  
print("not A2 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_1014 != a2_1014
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil)
```

**Penjelasan:**

2. Menampilkan Nilai Awal, mencetak nilai boolean dasar dari kedua variabel ke layar.

3. Operator Konjungsi (and), menghasilkan True hanya jika kedua input bernilai True.

4. Operator Disjungsi (or), menghasilkan True jika salah satu atau kedua input bernilai True.

5. Operator Negasi (not), membalikkan nilai boolean dasar (True menjadi False atau sebaliknya).

6. Disjungsi Eksklusif (XOR / !=), menghasilkan True hanya jika nilai kedua input berbeda.

### 2.2.6 Program "C_2611531014"

```python
# Buat file dengan nama Perbandingan_2611531014.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1014
# program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam python

angka1_1014 = int(input("input angka-1: "))
angka2_1014 = int(input("input angka-2: "))

# Lebih besar dari 
hasil = angka1_1014 > angka2_1014
print("\nOperator lebih besar dari")
print("angka1 > angka2 =", hasil)

# Lebih kecil dari
hasil = angka1_1014 < angka2_1014
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =", hasil)

# Lebih besar dari atau sama dengan 
hasil = angka1_1014 >= angka2_1014
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil)

# Lebih kecil dari atau sama dengan
hasil = angka1_1014 <= angka2_1014
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil)

# Sama dengan
hasil = angka1_1014 == angka2_1014
print("\nOperator sama dengan")
print("angka1 == angka2 =", hasil)

# Tidak sama dengan 
hasil = angka1_1014 != angka2_1014
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil)

# Tambahan: perbandingan berantai dalam python
hasil = 0 < angka1_1014 < 100
print("\nPerbandingan berantai")
print("0 < angka1 < 100 =", hasil)

hasil = 0 < angka2_1014 < 100
print("0 < angka2 < 100 =", hasil)
```

**Penjelasan:**
1. Input & Konversi Data, meminta input dua buah bilangan dari pengguna menggunakan fungsi input(), kemudian dikonversi langsung menjadi tipe data integer (int) dan disimpan ke variabel angka1_1014 serta angka2_1014.

2.
