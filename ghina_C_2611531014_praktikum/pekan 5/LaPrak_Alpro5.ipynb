## 2.1 Dasar Teori
1. Perulangan for pada Python,
Perulangan for adalah struktur perulangan yang digunakan untuk menjalankan suatu perintah berulang kali berdasarkan data atau rentang nilai yang diberikan. Dalam Python, for sering digunakan bersama fungsi range() untuk melakukan pengulangan dengan jumlah tertentu. Perulangan ini dapat digunakan untuk menampilkan angka, mengolah data, maupun membuat pola.

2. Fungsi range(),
range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan sebagai sumber pengulangan pada for. Bentuk sederhananya adalah range(stop), sedangkan untuk menentukan awal dan kenaikan dapat digunakan range(start, stop, step). Nilai stop tidak termasuk dalam urutan yang dihasilkan.

3. Nested Loop,
Nested loop adalah perulangan yang berada di dalam perulangan lainnya. Pada Python, for dapat ditempatkan di dalam for untuk melakukan proses yang lebih kompleks. Konsep ini berguna ketika program membutuhkan pengulangan berdasarkan baris dan kolom, misalnya dalam pembuatan pola angka atau karakter.

4. Penggunaan end="",
Pada fungsi print(), parameter end="" digunakan agar hasil yang dicetak tidak langsung berpindah ke baris berikutnya. Hal ini berguna ketika membuat pola karena beberapa karakter perlu dicetak dalam satu baris secara berurutan. Setelah seluruh isi satu baris selesai dicetak, print() tanpa argumen dapat digunakan untuk berpindah ke baris berikutnya.

5. Perulangan untuk Membuat Pola,
Perulangan for dapat digunakan untuk membuat pola dengan mengatur jumlah angka, karakter, dan spasi pada setiap baris. Jumlah pengulangan dapat dibuat berdasarkan nilai N yang dimasukkan pengguna sehingga ukuran pola dapat berubah secara dinamis. Dalam pembuatan pola jam pasir, perulangan digunakan untuk membentuk bagian atas, titik tengah, bagian bawah, serta bingkai pola.

6. Perulangan for pada Program Jam Pasir Kristal Palindromik,
Pada studi kasus praktikum ini, for digunakan untuk membentuk pola angka yang menurun dan menaik secara simetris. Nested loop digunakan untuk mencetak spasi, angka, dan garis pembatas. Nilai N menentukan ukuran pola, sedangkan perhitungan seperti 2 * (N - baris) digunakan untuk mengatur jarak atau posisi pola agar tetap simetris. Dengan demikian, program dapat menghasilkan pola yang berbeda sesuai dengan nilai N yang diberikan pengguna.

### 2.2.8 Program "C_2611531014"

```python
#Buat file dengan nama nested_for4_2611531014.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditam#bah 4 digit nim terakhir contoh: ulang_1014
# Program ini menggunakan fungsi input()

tinggi_1014 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1014 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1014 = tinggi_1014
    c_1014 = a_1014
    lebar_1014 = (2 * tinggi_1014) - 2

    for i_1014 in range(1, tinggi_1014 + 1):
        b_1014 = c_1014 + 1

        for j_1014 in range(1, lebar_1014 + 1):

            # Baris atas dan bawah
            if i_1014 == 1 or i_1014 == tinggi_1014:
                if j_1014 == 1 or j_1014 == lebar_1014:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_1014 == 1 or j_1014 == lebar_1014:
                    print("|", end="")
                elif j_1014 == c_1014:
                    print("<", end="")
                elif j_1014 == b_1014:
                    print(">", end="")
                elif j_1014 == lebar_1014 - c_1014:
                    print("<", end="")
                elif j_1014 == lebar_1014 - c_1014 + 1:
                    print(">", end="")
                elif b_1014 < j_1014 < lebar_1014 - c_1014:
                    print(".", end="")
                else:
                    print(" ", end="")

        print()

        a_1014 -= 2
        if a_1014 <= 0:
            c_1014 = (-a_1014) + 2
        else:
            c_1014 = a_1014
```

**Penjelasan:**
1. Bagian pertama kode ini berfungsi untuk menerima input tinggi pola dari pengguna, mengubahnya ke tipe data integer, serta melakukan validasi untuk memastikan nilai tinggi yang dimasukkan adalah bilangan genap.
2. Bagian kedua kode ini berfungsi untuk menginisialisasi variabel pendukung (a_1014, c_1014, lebar_1014) yang mengatur perhitungan dimensi serta posisi simbol dalam pola.
3. Bagian ketiga kode ini berfungsi untuk menjalankan perulangan luar (for i_1014) dan perulangan dalam (for j_1014) guna mencetak bingkai (menggunakan #, =, |) serta simbol hiasan di dalam pola (seperti <, >, .), dilanjutkan dengan pembaruan variabel kontrol posisi (c_1014) untuk membentuk pola simetris pada setiap baris.

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan for merupakan salah satu konsep penting dalam pemrograman Python yang digunakan untuk menjalankan perintah secara berulang sesuai dengan kebutuhan program. Dalam konteks pemrograman, konsep ini memanfaatkan perulangan for, fungsi range(), nested loop, dan parameter end="" sehingga program dapat melakukan proses secara teratur tanpa harus menuliskan perintah yang sama berulang kali.

Melalui pembuatan pola Jam Pasir Kristal Palindromik Berbingkai, dapat dipahami bahwa perulangan dapat digunakan untuk mengatur jumlah baris, posisi spasi, susunan angka, dan karakter pembatas. Penggunaan nested loop memungkinkan beberapa proses perulangan dilakukan secara bersamaan untuk membentuk pola yang lebih kompleks. Selain itu, penggunaan input N membuat ukuran pola dapat berubah secara dinamis sesuai dengan nilai yang diberikan pengguna. Dengan demikian, praktikum ini membantu meningkatkan pemahaman tentang cara kerja perulangan serta melatih kemampuan berpikir logis dan sistematis dalam membuat program Python.
