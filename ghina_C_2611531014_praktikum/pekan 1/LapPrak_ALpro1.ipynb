## 2.1 Dasar Teori

1.  Python adalah salah satu bahasa pemrograman yang digunakan untuk memberikan perintah kepada komputer. Python termasuk bahasa pemrograman yang cukup mudah dipelajari karena penulisan sintaksnya sederhana dan mudah dibaca.
Python dapat digunakan untuk berbagai macam kebutuhan, seperti membuat program sederhana, mengolah data, membuat website, kecerdasan buatan, machine learning, otomatisasi pekerjaan, hingga pengembangan aplikasi.
2. Visual Studio Code (VS Code) adalah aplikasi yang digunakan untuk menulis, mengedit, dan menjalankan kode program. VS Code bukanlah bahasa pemrograman, melainkan tempat atau alat yang digunakan programmer untuk membuat program. VS Code dapat digunakan untuk berbagai bahasa pemrograman, salah satunya Python. Dengan VS Code, programmer dapat membuat file program, mengatur folder proyek, menjalankan kode, serta melihat dan memperbaiki kesalahan pada program.
3. Git adalah sebuah sistem yang digunakan untuk mengatur dan mencatat perubahan yang terjadi pada suatu proyek atau kode program. Git sangat berguna ketika sebuah program terus mengalami perubahan karena setiap perubahan dapat dicatat dan dilihat kembali. Dengan Git, kita dapat mengetahui perkembangan program dari waktu ke waktu dan lebih mudah kembali ke versi sebelumnya jika terjadi kesalahan. Git biasanya digunakan melalui terminal atau fitur yang sudah tersedia di aplikasi seperti VS Code.
4. GitHub adalah layanan berbasis internet yang digunakan untuk menyimpan, mengelola, dan berbagi proyek pemrograman yang menggunakan Git. Jika Git digunakan untuk mengatur perubahan pada komputer, GitHub dapat menjadi tempat penyimpanan proyek tersebut secara online. Dengan GitHub, proyek dapat diakses dari perangkat lain dan dapat dibagikan kepada orang lain. GitHub juga banyak digunakan untuk bekerja sama dalam sebuah proyek karena beberapa orang dapat berkontribusi pada proyek yang sama.
5. Repository atau biasa disebut repo adalah tempat penyimpanan sebuah proyek yang dikelola menggunakan Git. Repository dapat berisi berbagai file yang diperlukan dalam sebuah proyek, seperti file kode Python, dokumen, gambar, dan file lainnya. Repository dapat berada di komputer sendiri yang disebut local repository, atau berada di layanan online seperti GitHub yang disebut remote repository. Jadi, repository dapat diibaratkan sebagai sebuah folder khusus yang berisi proyek sekaligus riwayat perubahan dari proyek tersebut.
6. Commit adalah proses mencatat perubahan yang telah dilakukan pada proyek ke dalam Git. Commit dapat dianggap seperti membuat sebuah titik penyimpanan atau checkpoint pada proyek. Setiap commit biasanya diberikan pesan yang menjelaskan perubahan yang dilakukan, misalnya "Menambahkan program perulangan" atau "Memperbaiki kesalahan pada kode". Dengan adanya commit, kita dapat melihat kembali perubahan yang sudah dilakukan dan mengetahui perkembangan proyek secara bertahap.
7. Push adalah proses mengirim perubahan yang sudah di-commit dari repository yang ada di komputer ke repository yang ada di GitHub. Misalnya, kita membuat program Python di VS Code, kemudian menggunakan Git untuk mencatat perubahan melalui commit. Setelah itu, kita melakukan push agar perubahan tersebut dikirim dan tersimpan di GitHub. Oleh karena itu, urutan sederhananya adalah membuat kode menggunakan Python, menulisnya di VS Code, mengelola perubahan menggunakan Git, menyimpannya dalam commit, kemudian melakukan push aga perubahan tersebut tersimpan di repository GitHub.
