## 2.1 Dasar Teori
1. Boolean: Tipe data logika yang hanya memiliki nilai true atau false, berfungsi sebagai landasan utama bagi program dalam mengevaluasi kondisi, mengambil keputusan (branching), serta mengatur jalurnya pengulangan (looping).

2. String: Tipe data yang menyimpan sekumpulan karakter berupa huruf, angka, simbol, maupun spasi, difungsikan khusus untuk mengolah dan menyajikan informasi berbentuk teks agar dapat dibaca serta dipahami oleh pengguna.

3. Konstanta: Tempat penyimpanan data di dalam memori yang nilainya dikunci dan bersifat read-only, berguna untuk mengamankan data krusial dari perubahan yang tidak disengaja selama program sedang dieksekusi.

4. Numerik: Tipe data khusus angka—baik berbentuk bilangan bulat (integer) maupun pecahan desimal (float)—yang dirancang untuk menjalankan berbagai jenis operasi aritmetika dan kalkulasi matematika secara akurat.

5. Input: Instruksi yang bertugas menangkap data atau perintah dari luar program secara real-time, membuat aplikasi menjadi lebih dinamis dan interaktif karena dapat merespons aksi yang diberikan oleh pengguna.

## 2.2 Kode Program

### 2.2.1 Program "C_2611531014"
Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan.

```python
#buat file dengan nama Boolean_2611451014.py
#Nama Variabel ditambah 4 digit nim terakhir contoh: nilai_1014
#Deklarasi variabel dengan tipe data Boolean
is_lulus = True
is_cumlaude = True

# Menggunakan Boolean 
nilai = 85
batas_lulus = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan = nilai >= batas_lulus #Hasilnya akan True

print ("=== Check Kelulusan ===")
print("Nilai:", nilai)
print("Apakah Lulus?:", status_kelulusan)
if is_lulus and is_cumlaude:
        print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan :**
1. Variabel Boolean (is_lulus, is_cumlaude)berfungsi untuk Menyimpan tipe data logika bernilai True untuk menandakan status kelulusan dan cumlaude.
2. Evaluasi Perbandingan (nilai >= batas_lulus)berfungsi untuk Membandingkan angka 85 dengan 75, sehingga status_kelulusan menghasilkan nilai True.
3. Output print() berfungsi untuk Menampilkan nilai angka 85 dan status kelulusan berupa nilai Boolean True di layar.

### 2.2.2 Program "C_2611531014"

```python
#buat file dengan nama Input_2611531014.py
#Buat program untuk menghitung luas persegi panjang
#Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1014
#Program ini menggunakan fungsi input() untuk meminta pengguna
#Memasukkan nilai panjang dan lebar persegi panjang
#Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_1014 = input('Masukkan nilai panjang:')
lebar_1014 = input('Masukkan nilai lebar:')
luas_1014 = int(panjang_1014) * int(lebar_1014)
print("Luas =", luas_1014)
```

**Penjelasan :**
1. Input Teks `input()` berfungsi untuk mengambil angka yang diketik pengguna dan menyimpannya sebagai *string* di `panjang_1014` dan `lebar_1014`.
* Konversi Angka `int()` berfungsi untuk mengubah teks tersebut menjadi bilangan bulat agar bisa dihitung.
* Hitung Luas berfungsi untuk mengalikan nilai panjang dan lebar, lalu menyimpannya di `luas_1014`.
* Cetak Hasil `print()` berfungsi untuk menampilkan teks `"Luas ="` beserta hasil perhitungannya ke layar.

### 2.2.3 Program "C_2611531014"

```python
#Buat file dengan nama Konstanta_2611531014.py
#Program ini menggunakan konstanta untuk menghitung luas lingkaran
#nama variabel ditambah 4 digit nim terakhir contoh: jari_1014

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_1014 = float(input('Masukkan nilai jari-jari: '))
luas_1014 = PI * jari_1014 * jari_1014
print("luas lingkaran dengan jari-jari %.2f adalah %.2f" %(jari_1014, luas_1014))
```

**Penjalasan :**
1. Deklarasi Konstanta `PI: Final = 3.14`, berfungsi untuk menetapkan `PI` sebagai nilai tetap yang nilainya tidak boleh diubah.
2. Cetak Nilai PI `print()`, berfungsi untuk menampilkan nilai `PI` menggunakan format angka desimal (`%f`).
3. Input Jari-jari `float(input(...))`, berfungsi untuk mengambil nilai dari pengguna dan mengonversinya menjadi angka desimal di `jari_1014`.
4. Hitung Luas berfungsi untuk Mengalikan `PI * jari_1014 * jari_1014` sesuai rumus $\pi r^2$ dan menyimpannya di `luas_1014`.
5. Cetak Hasil berfungsi untuk menampilkan hasil perhitungan dengan format 2 angka di belakang koma (`%.2f`).

### 2.2.4 Program "C_2611531014"

```python
#buat file dengan nama Numerik_2611531014.py
#nama variabel ditambah 4 digit terakhir NIM contoh: a_1014
#create a variable with numeric type value.
a_1014=100
print ("The type of variable having value",a_1014,"is",type(a_1014))

#create a variable with float value.
b_1014=10.2345
print("The type of variable having value",b_1014,"is",type(b_1014))

#create a variable with complex value.
c_1014=100+3j
print("The type of variable having value",c_1014,"is", type(c_1014))
```

*Penjelasan :**
1. Pada tipe data integer (`a = 100`), Variabel `a` menyimpan bilangan bulat. Fungsi `type(a)` memeriksa tipe datanya dan mengembalikan `<class 'int'>`.
2. Pada tipe data float (`b = 10.2345`), Variabel `b` menyimpan bilangan desimal (pecahan). Fungsi `type(b)` mengembalikan `<class 'float'>`.
3. Pada tipe data complex (`c = 100+3j`), Variabel `c` menyimpan bilangan kompleks yang terdiri dari bagian riil (`100`) dan imajiner (`3j`). Fungsi `type(c)` mengembalikan `<class 'complex'>`.
4. Pada Cetak Output (`print()`), berfungsi untuk menampilkan nilai variabel beserta tipe data hasil dari fungsi `type()`.
