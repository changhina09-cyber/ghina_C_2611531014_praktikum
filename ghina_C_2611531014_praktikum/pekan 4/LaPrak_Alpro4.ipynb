Dalam pemrograman, sebuah program tidak selalu menjalankan perintah secara berurutan. ## 2.1 Dasar Teori

1. if,
if digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True). Jika kondisi salah, perintah di dalam if tidak dijalankan. Contohnya, if umur >= 17: dapat digunakan untuk memeriksa apakah seseorang sudah berusia 17 tahun atau lebih.

2. if-else,
if-else digunakan ketika terdapat dua kemungkinan. Jika kondisi pada if benar, program menjalankan perintah pertama. Jika kondisi salah, program menjalankan perintah pada else. Struktur ini dapat digunakan untuk menentukan apakah seseorang memenuhi atau tidak memenuhi suatu syarat.

3. if-elif-else,
if-elif-else digunakan untuk memeriksa beberapa kondisi. Python akan memeriksa kondisi dari atas ke bawah dan menjalankan blok pertama yang kondisinya benar. Jika tidak ada kondisi yang terpenuhi, bagian else akan dijalankan.

4. Multi-if,
Multi-if merupakan penggunaan beberapa if secara terpisah. Berbeda dengan if-elif-else, setiap kondisi diperiksa secara independen sehingga lebih dari satu kondisi dapat terpenuhi. Pada praktikum ini, multi-if digunakan untuk menghitung diskon yang dapat ditambahkan atau diakumulasikan berdasarkan beberapa syarat.

5. match-case,
match-case digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan yang tersedia. Struktur ini cocok digunakan ketika program memiliki banyak pilihan berdasarkan satu nilai tertentu. Pada kasus sistem loket, match-case dapat digunakan untuk menentukan jenis wahana berdasarkan nomor paket yang dipilih pengunjung. case _ digunakan sebagai kondisi default apabila tidak ada pilihan yang sesuai.

6. Operator Logika,
Operator logika digunakan untuk menggabungkan atau membandingkan beberapa kondisi. Beberapa operator yang digunakan dalam percabangan adalah and, or, dan not. Selain itu, operator perbandingan seperti != dapat digunakan untuk memeriksa apakah dua nilai berbeda. Penggunaan operator tersebut membuat program dapat menangani kondisi yang lebih kompleks.

7. Penerapan dalam Program,
Struktur percabangan dapat digunakan dalam berbagai program, seperti menentukan kelayakan seseorang, memilih menu, memberikan diskon, menentukan kategori nilai, dan memvalidasi data. Dalam praktikum ini, percabangan diterapkan pada sistem loket wahana untuk menentukan paket wahana, memeriksa kelayakan pengunjung, menghitung diskon, dan menentukan status transaksi.

### 2.2.3 Program "C_2611531014"


```python
# Buat file dengan nama multi_if1_2611541014.py
# Buat program untuk kondisi if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1014
# Program ini menggunakan fungsi input()

umur_1014 = int(input("Masukkan umur Anda: "))
sim = input("Apakah Anda Sudah punya Sim C (y/t): ")[0]

if umur_1014 >= 17 and sim == "y":
    print("Anda Sudah Dewasa dan boleh mengendarai motor")

if umur_1014 >= 17 and sim != "y":
    print("Anda Sudah Dewasa tetapi tidak boleh mengendarai motor")

if umur_1014 < 17 and sim == "y":
    print("Anda belum cukup umur punya SIM")

if umur_1014 < 17 and sim != "y":
    print("Anda belum cukup umur bawa motor")
```

### 2.2.4 Program "C_2611531014"


```python
# Buat file dengan nama multi_if2_2611531014.py
# Buat program untuk kondisi if
# Nma variabel ditambah 4 digit nim terakhir contoh: ipk_1014
# Program ini menggunakan fungsi input()
# Program menghitung Diskon Belanja

# Input dari user
total_belanja_1014 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1014 = input("Apakah Anda member (y/t): ").strip().lower()
is_member_1014 = input_member_1014 in [ "y", "ya"]

#  Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya'
input_promo_1014 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1014 = input_promo_1014 in [ "y", "ya"]

total_diskon_persen_1014 = 0

# Multi-if terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_1014 > 1000000:
    total_diskon_persen_1014 += 10  # Diskon belanja besar

if is_member_1014:
    total_diskon_persen_1014 += 5  # Diskon member

if kode_promo_valid_1014:
    total_diskon_persen_1014 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_1014 = total_belanja_1014 * (total_diskon_persen_1014 / 100)
total_bayar_1014 = total_belanja_1014 - nominal_diskon_1014

# Output hasil 
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_1014}% (Rp{nominal_diskon_1014:,.0f})")
print(f"Total bayar : Rp{total_bayar_1014:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_1014}%")
# Output Total diskon yang anda dapatkan: 30% jika belanja > 1 juta , member dan kode promo valid
```

**Penjelasan:**
1. Bagian pertama kode ini berfungsi untuk mengambil masukan total belanja dari pengguna dalam bentuk bilangan desimal (float) melalui input() dan menyimpannya ke variabel total_belanja_1014.
2. Bagian kedua kode ini berfungsi untuk mengambil masukan status member, menghapus spasi (.strip()), mengubah teks menjadi huruf kecil (.lower()), serta mengecek apakah nilainya "y" atau "ya" untuk menghasilkan nilai boolean pada is_member_1014.
3. Bagian ketiga kode ini berfungsi untuk mengambil masukan keabsahan kode promo dengan pemrosesan teks yang sama untuk menghasilkan nilai boolean pada kode_promo_valid_1014.
4. Bagian keempat kode ini berfungsi untuk menginisialisasi variabel total_diskon_persen_1014 dengan nilai dasar 0 sebagai penampung persentase diskon.
5. Bagian kelima kode ini berfungsi untuk memeriksa secara mandiri apakah total belanja > 1.000.000, lalu menambahkan diskon sebesar 10\% jika kondisi terpenuhi.
6. Bagian keenam kode ini berfungsi untuk memeriksa secara mandiri apakah pengguna merupakan member, lalu menambahkan diskon sebesar 5\% jika bernilai benar (true).
7. Bagian ketujuh kode ini berfungsi untuk memeriksa secara mandiri apakah kode promo valid, lalu menambahkan diskon sebesar 15\% jika bernilai benar (true).
8. Bagian kedelapan kode ini berfungsi untuk menghitung nominal diskon dalam rupiah (nominal_diskon_1014) serta total pembayaran akhir (total_bayar_1014) berdasarkan akumulasi diskon.
9. Bagian kesembilan kode ini berfungsi untuk menampilkan rincian pembayaran ke layar, yang mencakup persentase diskon, nominal diskon, dan total biaya yang harus dibayar.

### 2.2.6 Program "C_2611531014"


```python
# Buat file dengan nama if_2611541014.py
# Buat program untuk kondisi if
# Nma variabel ditambah 4 digit nim terakhir contoh: ipk_1014
# Program konversi nilai angka menjadi huruf

nilai_1014 = int(input("Masukkan nilai angka = "))

if nilai_1014 >= 81:
    print("A")
elif nilai_1014 >= 70:
    print("B")
elif nilai_1014 >= 60:
    print("C")
elif nilai_1014 >= 50:
    print("D")
else:
    print("E")
    
```

4. Bagian keempat kode ini berfungsi untuk mengecek kondisi alternatif kedua (elif), yaitu jika nilai berada pada rentang $60 \le \text{nilai\_1014} < 70$, maka program akan mencetak nilai huruf "C".
5. Bagian kelima kode ini berfungsi untuk mengecek kondisi alternatif ketiga (elif), yaitu jika nilai berada pada rentang $50 \le \text{nilai\_1014} < 60$, maka program akan mencetak nilai huruf "D".
6. Bagian keenam kode ini berfungsi untuk menangani kondisi terakhir (else), yaitu jika nilai $< 50$, maka program secara otomatis akan mencetak nilai huruf "E".

### 2.2.7 Program "C_2611531014"


```python
# Buat file dengan nama match_case1.py
# Buat program untuk kondisi if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1014
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_1014 = int(input("Masukkan angka bulan (1-12):"))

match bulan_1014:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasn:**
1. Bagian pertama kode ini berfungsi untuk mengambil masukan dari pengguna berupa angka bulan (1–12), mengonversinya menjadi tipe data bilangan bulat (integer), lalu menyimpannya ke dalam variabel bulan_1014.
2. Bagian kedua kode ini berfungsi untuk mencocokkan nilai variabel bulan_1014 dengan pilihan kasus (case) dari angka 1 hingga 12 menggunakan struktur match-case, di mana setiap case bertugas mencetak nama bulan yang sesuai (misalnya case 1 untuk "Januari", case 2 untuk "Februari", hingga case 12 untuk "Desember").
3. Bagian ketiga kode ini berfungsi untuk menangani kondisi default atau wildcard (case _), yang akan mengeksekusi perintah mencetak "Angka tidak valid" jika pengguna memasukkan angka di luar rentang 1 hingga 12.

```python
# tugas pekan 4 
# Struktur percabangan python

print("=== SISTEM LOKET ALPRO ADVANTURE PARK ===")

#1. INPUT DATA PENGUNJUNG
nama_1014 = input("Masukkan nama pengunjung :")
umur_1014 = int(input("input umur anda :"))
sim_1014 = input("Apakah anda sudah punya SIM C (y/t):") [0].lower()

print("\nPilihan paket Wahana (1-5):")
print(" 1. Safari Rimba (Rp 50,000)")
print(" 2. Arung Jeram  (Rp 75,000)")
print(" 3. Motor ATV Ekstrim (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp100,000)")
print(" 5. All-Accsess VIP (RP 220,000)")

paket_1014 = int(input("masukkan nomor paket (1-5) :"))
jumlah_tiket_1014 = int(input("masukkan jumlah tiket :" ))

# 2. IF TUNGGAL
#validasi jumlah tiket

if jumlah_tiket_1014 <= 0:
    print("Peringatan: Kuota tiket tidak valid")
    print("Program selesai.")
    exit()

# 3. MATCH-CASE
#menentukan jenis wahana dan harga

match paket_1014:
    case 1:
        nama_wahana_1014 = "Wahana Safari Rimba"
        harga_satuan_1014 = 50000
    case 2: 
        nama_wahana_1014 = "Wahana Arung Jeram"
        harga_satuan_1014 = 75000
    case 3:
        nama_wahana_1014 = "Wahana Motot ATV Ekstrim"
        harga_satuan_1014 = 120000
    case 4:
        nama_wahana_1014 = "Wahana Roller Coaster Kilat"
        harga_satuan_1014 = 100000
    case 5: 
        nama_wahana_1014 = "Wahana All_Accsess VIP"
        harga_satuan_1014 = 220000
    case _:
        print("paket wahana tidak valid")
        print("Program selesai.")
        exit()

# INPUT MEMBER DAN KODE PROMO
is_member_1014 = input("Apakah anda member? (y/t) :").strip().lower()
kode_promo_valid_1014 = input("Apakah kode promo valid? (y/t) :").strip().lower()

# 5.VALIDASI IZIN KENDALI WAHANA
#khusus paket 3 menggunakan if-elif-else

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
if paket_1014 == 3:
    if umur_1014 >= 17 and sim_1014 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_1014 >= 17 and sim_1014 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh" "bawa motor ATV (wajib didampingi instruktur).")
    elif umur_1014 < 17 and sim_1014 == 'y':
        print("Status Akses: identitas tidak valid:" "belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur" "dan tidak boleh bawa motor ATV.")

# Untuk paket selain 3 menggunakan if-else sederhana
else:
    if umur_1014 >= 10:
        print("Status Akses: Anda memenuhi batas usia wahana.")
    else:
        print("Status Akses: Anda belum memenuhi batas usia wahana.")

# 6. PERHITUNGAN SUBTOTAL
subtotal_1014 = harga_satuan_1014 * jumlah_tiket_1014

#variabel awal untuk menampung total diskon
total_diskon_persen_1014 = 0

# 7. MULTI-IF
#setiap if berdiri sendiri agar diskon dapat ditumpuk

#diskon belanja besar
if subtotal_1014 >= 200000:
    total_diskon_persen_1014 += 10
#diskon member
if is_member_1014 in ['y', 'ya']:
    total_diskon_persen_1014 += 5
#diskon voucher promo
if kode_promo_valid_1014 in ['y', 'ya']:
    total_diskon_persen_1014 += 15
#diskon tambahan rombongan
if jumlah_tiket_1014 >= 5:
    total_diskon_persen_1014 += 5

# 8. PERHITUNGAN TOTAL PEMBAYARAN
nominal_diskon_1014 = (subtotal_1014 * (total_diskon_persen_1014 / 100))
total_bayar_1014 = subtotal_1014 - nominal_diskon_1014

# 9. IF-ELSE UNTUK AUDIT TRANSAKSI
if total_bayar_1014 > 300000:
    catatan_layanan_1014 = ("Selamat! Anda berhak mendapatkan souvenir gratis.")
else:
    catatan_layanan_1014 = "Terima kasih telah berkunjung."

# 10. MENAMPILKAN RINCIAN PEMBAYARAN
print("\n--- RINCIAN PEMBAYARAN ---")
print(f"Nama Pengunjung : {nama_1014}")
print(f"Wahana : {nama_wahana_1014}")
print(f"Jumlah Tiket : {jumlah_tiket_1014}")
print(f"Harga Satuan : Rp {harga_satuan_1014}")
print(f"Subtotal Belanja : Rp {subtotal_1014:,.0f}")
print(f"Total Diskon : {total_diskon_persen_1014}%" f"(Rp {nominal_diskon_1014:,.0f})")
print(f"Total Bayar : Rp {total_bayar_1014:,.0f}")
print(f"Catatan Layanan : {catatan_layanan_1014}")

print("\nProgram Selesai")
```

Dalam konteks pemrograman, konsep ini memanfaatkan "if", "if-else", "if-elif-else", "multi-if", dan "match-case" sehingga program dapat menjalankan perintah yang berbeda sesuai dengan keadaan yang terjadi. Selain itu, penggunaan multi-if memungkinkan beberapa jenis diskon dijumlahkan apabila syaratnya terpenuhi.
